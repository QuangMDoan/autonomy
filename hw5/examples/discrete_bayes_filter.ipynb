```python
def transition_model(prev_x, u_t, x_t): # return p(x_t | x_t-1, u_t)
    model = { 
        ('open',    'push', 'open')     : 1.0,
        ('open',    'push', 'closed')   : 0.0,
        ('closed',  'push', 'open')     : 0.8,
        ('closed',  'push', 'closed')   : 0.2,

        ('open',    'nothing', 'open')      : 1.0,
        ('open',    'nothing', 'closed')    : 0.0,
        ('closed',  'nothing', 'open')      : 0.0,
        ('closed',  'nothing', 'closed')    : 1.0
    }
    return model[(prev_x, u_t, x_t)]

def measurement_model(x_t, z_t): # return p(z_t | x_t)
    model = { 
        ('open',    'open')     : 0.6,
        ('open',    'closed')   : 0.4,
        ('closed',  'open')     : 0.2,
        ('closed',   'closed')   : 0.8
    }
    return model[(x_t, z_t)]

def given():
    print("Given")
    print("p(x0=open) = 0.5")
    print("p(x0=closed) = 0.5")

u1 = "nothing"
z1 = "open"

def question():
    print("Question: ")
    print("what is my belief that x1 = open?")

def pred_belief(t, ut, xt):
    sum = 0
    for xi in ["open", "closed"]:
        transition_p = transition_model(xi, ut, xt)
        prev_belief = belief(t-1, xi, ut)
        sum += transition_p * prev_belief
    return sum
        
def belief(t, xt, ut, zt=None):
    priors = {
        "open" :   0.5,
        "closed" : 0.5
    }

    if t == 0: return priors[xt]
    
    predict_belief = {
        'open' : measurement_model("open", zt) * pred_belief(t, ut, 'open'),
        'closed' : measurement_model("closed", zt) * pred_belief(t, ut, 'closed')
    }

    predict_belief_sum          = predict_belief["open"] + predict_belief["closed"]
    predict_belief["open"]      = predict_belief["open"] / predict_belief_sum
    predict_belief["closed"]    = predict_belief["closed"] / predict_belief_sum

    return predict_belief[xt]

# what is my belief that x1 = open?
# open_x1_belief = belief(1, 'open',  u1, z1)
# print(f"open_x1_belief: {open_x1_belief :.2f}")
```
